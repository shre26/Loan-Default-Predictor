```python
import sys
sys.path.append("../src")
from data_preprocessing import load_data, get_train_test_split, encode_binary_features, build_preprocessor
import joblib
```

```python
df = load_data()
```

Split before any scaling/encoding (Leakage-Prevention Step)

```python
X_train, X_test, y_train, y_test = get_train_test_split(df)
```

```python
X_train, X_test = encode_binary_features(X_train, X_test)
```

```python
preprocessor = build_preprocessor()
```

```python
X_train_processed = preprocessor.fit_transform(X_train)   # fit ONLY on train
X_test_processed = preprocessor.transform(X_test)         # transform only, no fit
```

```python
joblib.dump(preprocessor, "../models/preprocessor.pkl")
joblib.dump((X_train_processed, X_test_processed, y_train, y_test), "../data/processed/train_test_split.pkl")
```

```python
print("Train shape:",X_train_processed.shape)
print("Test shape:",X_test_processed.shape)
print(y_train.value_counts(normalize=True))
```
